```python
# Load Jane Austen's Emma
text = gutenberg.raw('austen-emma.txt')

print(text[:1000])
```

```python
from nltk.tokenize import word_tokenize

# Convert text to lowercase
text = text.lower()

# Tokenize the text
gutenberg_tokens = word_tokenize(text)

# Keep only alphabetic words
gutenberg_tokens = [
    word for word in gutenberg_tokens
    if word.isalpha()
]

print("First 30 tokens:")
print(gutenberg_tokens[:30])

print("\nTotal tokens:")
print(len(gutenberg_tokens))
```

```python
# Unigram frequency
q2_unigram_freq = FreqDist(gutenberg_tokens)

# Generate bigrams
q2_bigrams = list(ngrams(gutenberg_tokens, 2))

# Bigram frequency
q2_bigram_freq = FreqDist(q2_bigrams)

print("Vocabulary Size:", len(q2_unigram_freq))
print("Total Unigrams:", len(gutenberg_tokens))
print("Total Bigrams:", len(q2_bigrams))

print("\nTop 10 Unigrams:")
print(q2_unigram_freq.most_common(10))

print("\nTop 10 Bigrams:")
print(q2_bigram_freq.most_common(10))
```

```python
# Vocabulary size
V = len(q2_unigram_freq)

def bigram_probability(word1, word2):
    
    bigram_count = q2_bigram_freq[(word1, word2)]
    unigram_count = q2_unigram_freq[word1]

    probability = (bigram_count + 1) / (unigram_count + V)

    return probability
```

```python
print("P(was | emma):", bigram_probability("emma", "was"))
```

```python
def sentence_score(sentence):

    # Normalize
    sentence = sentence.lower()

    # Tokenize
    words = word_tokenize(sentence)

    # Keep alphabetic words
    words = [word for word in words if word.isalpha()]

    # Start log probability
    log_probability = 0

    # Calculate probability of each bigram
    for word1, word2 in ngrams(words, 2):

        probability = bigram_probability(word1, word2)

        log_probability += math.log(probability)

    return log_probability
```

```python
test_sentences = [
    "Emma was happy",
    "The house was beautiful",
    "The cat sat on the table"
]

for sentence in test_sentences:

    score = sentence_score(sentence)

    print("Sentence:", sentence)
    print("Log Probability Score:", score)
    print()
```

Smoothing is important because some valid word combinations may not appear
in the training corpus.

Without smoothing, an unseen bigram receives probability zero, which makes
the probability of the complete sentence zero.

Add-one smoothing assigns a small non-zero probability to unseen bigrams.
Therefore, smoothing improves the ability of the language model to handle
unseen data and generalize better.

```python
# question 3
tagged_sentences = brown.tagged_sents(tagset='universal')

print("First tagged sentence:")
print(tagged_sentences[0])

print("\nTotal tagged sentences:")
print(len(tagged_sentences))
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    tagged_sentences,
    test_size=0.2,
    random_state=42
)

print("Total sentences:", len(tagged_sentences))
print("Training sentences:", len(train_data))
print("Testing sentences:", len(test_data))
```

```python
from nltk.tag import hmm

# Create HMM trainer
trainer = hmm.HiddenMarkovModelTrainer()

# Train the model
hmm_tagger = trainer.train_supervised(train_data)

print("HMM model trained successfully!")
```

```python
unseen_sentences = [
    "The dog is running",
    "She likes natural language processing",
    "Students are studying in the library",
    "The computer works very fast",
    "Machine learning is interesting"
]

for sentence in unseen_sentences:

    words = word_tokenize(sentence)

    tagged = hmm_tagger.tag(words)

    print("\nSentence:")
    print(sentence)

    print("POS Tags:")
    print(tagged)
```

```python
sample_test_data = test_data[:500]

accuracy = hmm_tagger.accuracy(sample_test_data)

print("HMM Test Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

Hidden States:
The hidden states are the Part-of-Speech tags such as NOUN, VERB, ADJ,
ADV and DET.

Observations:
The observations are the words in the sentence.

Transition Probability:
It represents the probability of moving from one POS tag to another.
For example:

P(NOUN | DET)

represents the probability that a noun follows a determiner.

Emission Probability:
It represents the probability of observing a word given a POS tag.
For example:

P(dog | NOUN)

represents the probability of the word "dog" being generated by the
NOUN state.

```python
from sklearn.datasets import fetch_20newsgroups

categories = [
    'sci.space',
    'comp.graphics'
]

newsgroups = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

print("Categories:")
print(newsgroups.target_names)

print("\nNumber of documents:")
print(len(newsgroups.data))
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create vectorizer
vectorizer = CountVectorizer(
    stop_words='english'
)

# Convert text into term-frequency matrix
X = vectorizer.fit_transform(newsgroups.data)

# Get all terms
terms = vectorizer.get_feature_names_out()

# Calculate frequencies
frequencies = np.asarray(X.sum(axis=0)).flatten()

# Create DataFrame
terms_df = pd.DataFrame({
    "Term": terms,
    "Frequency": frequencies
})

# Sort by frequency
terms_df = terms_df.sort_values(
    by="Frequency",
    ascending=False
)

# Display top 20
print("Top 20 Frequent Domain Terms:")
print(terms_df.head(20))
```

```python
space_data = fetch_20newsgroups(
    subset='train',
    categories=['sci.space'],
    remove=('headers', 'footers', 'quotes')
)

vectorizer_space = CountVectorizer(
    stop_words='english'
)

X_space = vectorizer_space.fit_transform(space_data.data)

space_terms = vectorizer_space.get_feature_names_out()

space_frequencies = np.asarray(
    X_space.sum(axis=0)
).flatten()

space_df = pd.DataFrame({
    "Term": space_terms,
    "Frequency": space_frequencies
})

space_df = space_df.sort_values(
    by="Frequency",
    ascending=False
)

print("TOP 20 SCI.SPACE TERMS:")
print(space_df.head(20))
```

```python
graphics_data = fetch_20newsgroups(
    subset='train',
    categories=['comp.graphics'],
    remove=('headers', 'footers', 'quotes')
)

vectorizer_graphics = CountVectorizer(
    stop_words='english'
)

X_graphics = vectorizer_graphics.fit_transform(
    graphics_data.data
)

graphics_terms = vectorizer_graphics.get_feature_names_out()

graphics_frequencies = np.asarray(
    X_graphics.sum(axis=0)
).flatten()

graphics_df = pd.DataFrame({
    "Term": graphics_terms,
    "Frequency": graphics_frequencies
})

graphics_df = graphics_df.sort_values(
    by="Frequency",
    ascending=False
)

print("TOP 20 COMP.GRAPHICS TERMS:")
print(graphics_df.head(20))
```

Ambiguity affects NLP because the same word can have multiple meanings.
The correct meaning can only be identified by analyzing the context.

Coreference affects NLP because pronouns such as he, she, it, and they
must be connected to their correct antecedents.

Domain-specific vocabulary also affects NLP performance because different
domains use specialized terminology. A model trained on general language
may not correctly understand technical vocabulary from domains such as
space science or computer graphics.

Therefore, context, coreference resolution, and domain knowledge are
important for improving NLP system performance.
