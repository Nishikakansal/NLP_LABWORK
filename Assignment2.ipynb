```python
# Q1: Exploring WordNet Synsets and Meanings

import nltk
from nltk.corpus import wordnet as wn

# Download required NLTK resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Select 10 English words
words = [
    "car",
    "bank",
    "bat",
    "dog",
    "light",
    "plant",
    "book",
    "teacher",
    "student",
    "computer"
]

# Display synsets, definitions and examples
for word in words:
    print("=" * 70)
    print("WORD:", word)

    synsets = wn.synsets(word)

    # Display selected synsets
    for synset in synsets[:5]:
        print("\nSynset:", synset.name())
        print("Definition:", synset.definition())

        examples = synset.examples()

        if examples:
            print("Example:", examples[0])
        else:
            print("Example: No example available")

# Display words having multiple meanings
print("\n\nWORDS WITH MULTIPLE SYNSETS")
print("=" * 70)

for word in words:
    synsets = wn.synsets(word)

    if len(synsets) > 1:
        print("\n", word, "has", len(synsets), "synsets.")

        for synset in synsets[:3]:
            print("Synset:", synset.name())
            print("Meaning:", synset.definition())
```

```python
# Q2: Hypernyms and Semantic Hierarchy

import nltk
import pandas as pd
from nltk.corpus import wordnet as wn

# Download WordNet
nltk.download('wordnet')
nltk.download('omw-1.4')

# Select 10 nouns
words = [
    "dog",
    "cat",
    "car",
    "computer",
    "teacher",
    "student",
    "book",
    "tree",
    "bird",
    "house"
]

data = []

for word in words:

    # Get noun synsets
    synsets = wn.synsets(word, pos=wn.NOUN)

    if not synsets:
        continue

    # Select the first noun synset
    synset = synsets[0]

    # Direct hypernyms
    direct_hypernyms = synset.hypernyms()

    direct_names = [
        h.name() for h in direct_hypernyms
    ]

    # Trace hierarchy for two or more levels
    chain = [synset.name()]

    current = synset

    for i in range(3):
        hypernyms = current.hypernyms()

        if not hypernyms:
            break

        current = hypernyms[0]
        chain.append(current.name())

    data.append({
        "Word": word,
        "Synset": synset.name(),
        "Direct Hypernyms": ", ".join(direct_names),
        "Hypernym Chain": " -> ".join(chain)
    })

# Create DataFrame
df = pd.DataFrame(data)

print(df.to_string(index=False))
```

```python
# Q3: WordNet Semantic Similarity

import nltk
import pandas as pd
from nltk.corpus import wordnet as wn

# Download WordNet
nltk.download('wordnet')
nltk.download('omw-1.4')

# Word pairs
pairs = [
    ("car", "automobile"),
    ("dog", "cat"),
    ("dog", "animal"),
    ("cat", "animal"),
    ("car", "vehicle"),
    ("teacher", "student"),
    ("dog", "computer"),
    ("book", "computer")
]

results = []

for word1, word2 in pairs:

    # Get noun synsets
    synsets1 = wn.synsets(word1, pos=wn.NOUN)
    synsets2 = wn.synsets(word2, pos=wn.NOUN)

    if not synsets1 or not synsets2:
        continue

    # Select the first suitable synset
    synset1 = synsets1[0]
    synset2 = synsets2[0]

    # Calculate path similarity
    similarity = synset1.path_similarity(synset2)

    results.append({
        "Word 1": word1,
        "Word 2": word2,
        "Synset 1": synset1.name(),
        "Synset 2": synset2.name(),
        "Path Similarity": similarity
    })

# Create DataFrame
df_similarity = pd.DataFrame(results)

# Rank from highest to lowest similarity
df_similarity = df_similarity.sort_values(
    by="Path Similarity",
    ascending=False
).reset_index(drop=True)

print(df_similarity.to_string(index=False))
```

```python
# Q4: Homonyms and Contextual Meaning

import nltk
import pandas as pd
from nltk.corpus import wordnet as wn

# Download WordNet
nltk.download('wordnet')
nltk.download('omw-1.4')

# Manually created contextual sentences
examples = [
    {
        "Word": "bank",
        "Sentence": "I deposited money in the bank.",
        "Meaning": "Financial institution",
        "Contextual Clue": "deposited money"
    },
    {
        "Word": "bank",
        "Sentence": "We sat on the bank of the river.",
        "Meaning": "Side of a river",
        "Contextual Clue": "river"
    },

    {
        "Word": "bat",
        "Sentence": "The bat flew out of the cave.",
        "Meaning": "Flying mammal",
        "Contextual Clue": "flew, cave"
    },
    {
        "Word": "bat",
        "Sentence": "He hit the ball with a wooden bat.",
        "Meaning": "Sports equipment",
        "Contextual Clue": "hit the ball"
    },

    {
        "Word": "bark",
        "Sentence": "The dog began to bark loudly.",
        "Meaning": "Sound made by a dog",
        "Contextual Clue": "dog, loudly"
    },
    {
        "Word": "bark",
        "Sentence": "The bark of the tree was rough.",
        "Meaning": "Outer covering of a tree",
        "Contextual Clue": "tree, rough"
    },

    {
        "Word": "light",
        "Sentence": "The light from the lamp was bright.",
        "Meaning": "Illumination",
        "Contextual Clue": "lamp, bright"
    },
    {
        "Word": "light",
        "Sentence": "This bag is very light.",
        "Meaning": "Not heavy",
        "Contextual Clue": "bag, not heavy"
    },

    {
        "Word": "match",
        "Sentence": "He lit the candle with a match.",
        "Meaning": "Small stick used to produce fire",
        "Contextual Clue": "lit the candle"
    },
    {
        "Word": "match",
        "Sentence": "The football match started at 6 PM.",
        "Meaning": "Sports competition",
        "Contextual Clue": "football, started"
    }
]

# Display WordNet synsets for each homonym
homonyms = ["bank", "bat", "bark", "light", "match"]

print("WORDNET SYNSETS FOR HOMONYMS")
print("=" * 70)

for word in homonyms:

    print("\nWord:", word)

    synsets = wn.synsets(word)

    for synset in synsets[:5]:
        print("Synset:", synset.name())
        print("Definition:", synset.definition())

# Create final DataFrame
df_context = pd.DataFrame(examples)

print("\n\nFINAL CONTEXTUAL MEANING TABLE")
print("=" * 70)

print(df_context.to_string(index=False))
```
